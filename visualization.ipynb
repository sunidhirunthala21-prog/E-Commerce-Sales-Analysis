```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_style("whitegrid")

# Display all columns
pd.set_option('display.max_columns', None)

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("ecommerce_analysis_dataset.csv")
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
df.describe()
```

```python
df.head(10)
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

# Convert Order_Date into datetime format

df['Order_Date'] = pd.to_datetime(df['Order_Date'])

df['Month'] = df['Order_Date'].dt.month
df['Month_Name'] = df['Order_Date'].dt.month_name()
df['Year'] = df['Order_Date'].dt.year

print("\nData cleaning completed!")
```

```python
# Filter delivered orders

delivered_df = df[df['Order_Status'] == 'Delivered'].copy()

# Calculate important metrics

total_orders = df['Order_ID'].nunique()

delivered_orders = delivered_df['Order_ID'].nunique()

total_customers = df['Customer_ID'].nunique()

total_revenue = delivered_df['Total_Amount_INR'].sum()

total_products_sold = delivered_df['Quantity'].sum()

average_order_value = total_revenue / delivered_orders

# Display results

print("========== E-COMMERCE SALES OVERVIEW ==========")

print("Total Orders:", total_orders)

print("Delivered Orders:", delivered_orders)

print("Total Customers:", total_customers)

print("Total Revenue: ₹", round(total_revenue, 2))

print("Total Products Sold:", total_products_sold)

print("Average Order Value: ₹", round(average_order_value, 2))
```

```python
# Filter delivered orders

delivered_df = df[df['Order_Status'] == 'Delivered'].copy()

# Calculate important metrics

total_orders = df['Order_ID'].nunique()

delivered_orders = delivered_df['Order_ID'].nunique()

total_customers = df['Customer_ID'].nunique()

total_revenue = delivered_df['Total_Amount_INR'].sum()

total_products_sold = delivered_df['Quantity'].sum()

average_order_value = total_revenue / delivered_orders

# Display results

print("========== E-COMMERCE SALES OVERVIEW ==========")

print("Total Orders:", total_orders)

print("Delivered Orders:", delivered_orders)

print("Total Customers:", total_customers)

print("Total Revenue: ₹", round(total_revenue, 2))

print("Total Products Sold:", total_products_sold)

print("Average Order Value: ₹", round(average_order_value, 2))
```

```python
# Create sales overview dashboard

metrics = {
    'Total Orders': total_orders,
    'Delivered Orders': delivered_orders,
    'Total Customers': total_customers,
    'Products Sold': total_products_sold
}

plt.figure(figsize=(10, 5))

plt.bar(metrics.keys(), metrics.values())

plt.title("E-Commerce Business Overview", fontsize=16)

plt.xlabel("Business Metrics")

plt.ylabel("Count")

plt.xticks(rotation=15)

plt.tight_layout()

plt.show()
```

```python
# Calculate category-wise revenue

category_sales = delivered_df.groupby(
    'Product_Category'
)['Total_Amount_INR'].sum().sort_values(ascending=False)

print(category_sales)

# Plot category-wise sales

plt.figure(figsize=(10, 6))

category_sales.plot(kind='bar')

plt.title("Category-wise Revenue Analysis", fontsize=16)

plt.xlabel("Product Category")

plt.ylabel("Revenue (INR)")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
# Calculate monthly revenue

monthly_sales = delivered_df.groupby(
    'Month'
)['Total_Amount_INR'].sum()

# Plot monthly sales

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o',
    linewidth=2
)

plt.title("Monthly Sales Trend", fontsize=16)

plt.xlabel("Month")

plt.ylabel("Revenue (INR)")

plt.xticks(range(1, 13))

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
# Calculate quantity sold by product

top_products = delivered_df.groupby(
    'Product'
)['Quantity'].sum().sort_values(ascending=False).head(10)

print("Top 10 Best-Selling Products:")
print(top_products)

# Plot top products

plt.figure(figsize=(12, 6))

top_products.sort_values().plot(kind='barh')

plt.title("Top 10 Best-Selling Products", fontsize=16)

plt.xlabel("Quantity Sold")

plt.ylabel("Product Name")

plt.tight_layout()

plt.show()
```

```python
# Count customers by gender

gender_count = df.groupby('Gender')['Customer_ID'].nunique()

print(gender_count)

# Create pie chart

plt.figure(figsize=(7, 7))

plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Customer Distribution by Gender", fontsize=16)

plt.show()
```

```python
# Calculate state-wise revenue

state_sales = delivered_df.groupby(
    'State'
)['Total_Amount_INR'].sum().sort_values(ascending=False)

print("State-wise Sales:")
print(state_sales)

# Plot state-wise revenue

plt.figure(figsize=(12, 6))

state_sales.plot(kind='bar')

plt.title("State-wise Revenue Analysis", fontsize=16)

plt.xlabel("State")

plt.ylabel("Revenue (INR)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# Count payment methods

payment_count = df['Payment_Method'].value_counts()

print("Payment Method Analysis:")
print(payment_count)

# Plot payment methods

plt.figure(figsize=(10, 6))

sns.barplot(
    x=payment_count.index,
    y=payment_count.values
)

plt.title("Payment Method Analysis", fontsize=16)

plt.xlabel("Payment Method")

plt.ylabel("Number of Orders")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
# Count order statuses

status_count = df['Order_Status'].value_counts()

print("Order Status Analysis:")
print(status_count)

# Create pie chart

plt.figure(figsize=(8, 8))

plt.pie(
    status_count,
    labels=status_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Order Status Distribution", fontsize=16)

plt.show()
```

```python
# Select numerical columns

numerical_data = delivered_df[
    ['Age', 'Quantity', 'Unit_Price_INR',
     'Discount_Percent', 'Total_Amount_INR']
]

# Calculate correlation

correlation = numerical_data.corr()

# Plot heatmap

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap", fontsize=16)

plt.tight_layout()

plt.show()
```

```python
# Find highest revenue category

best_category = category_sales.idxmax()

# Find best-selling product

best_product = top_products.idxmax()

# Find highest revenue state

best_state = state_sales.idxmax()

# Find most popular payment method

popular_payment = payment_count.idxmax()

# Find highest sales month

best_month = monthly_sales.idxmax()

print("========== FINAL E-COMMERCE ANALYSIS REPORT ==========")

print("Total Orders:", total_orders)

print("Delivered Orders:", delivered_orders)

print("Total Customers:", total_customers)

print("Total Revenue: ₹", round(total_revenue, 2))

print("Average Order Value: ₹", round(average_order_value, 2))

print("\n========== KEY FINDINGS ==========")

print("Highest Revenue Category:", best_category)

print("Best Selling Product:", best_product)

print("Highest Revenue State:", best_state)

print("Most Popular Payment Method:", popular_payment)

print("Highest Sales Month:", 
      df.loc[df['Month'] == best_month, 'Month_Name'].iloc[0])

print("\nAnalysis Completed Successfully!")
```

```python
# Save cleaned dataset

df.to_csv("cleaned_ecommerce_dataset.csv", index=False)

# Save category-wise sales

category_sales.to_csv("category_sales.csv")

# Save monthly sales

monthly_sales.to_csv("monthly_sales.csv")

# Save state-wise sales

state_sales.to_csv("state_sales.csv")

print("All analysis files saved successfully!")
```
